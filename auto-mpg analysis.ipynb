```python
import seaborn as sns #alternative of matplotlib 
```

```python
sns.boxplot(data['displacement'])
#IQR(Inter Quartile Range(25% to 75%)) which measures the spread of the middle of 50% of data and is useful for
#outlier detection
```

```python
sns.boxplot(data['acceleration'])#these 6 circles are outliers
```

```python
data['acceleration'].describe()
#here quartile range of Q1(lower line) is 8 and of Q3(higher line) is 24.8
```

```python
# To find IQR value manually
q3=data['acceleration'].quantile(0.75)
q1=data['acceleration'].quantile(0.25)
IQR=q3-q1
print(IQR)
```

```python
high=q3+(1.5*IQR)#1.5 is the margin error
low=q1-(1.5*IQR)
print(high)#the value is more accurate than maxvalue in describe method
print(low)
```

```python
data[(data['acceleration']<=low) | (data['acceleration']>=high)] # To find outliers
```

```python
data[(data['acceleration']>=low) & (data['acceleration']<=high)] # To remove outliers
```
